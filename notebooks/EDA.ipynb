```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(14, 10))
import scipy.stats as stats
sn.boxplot(x=df["monthly_emi_usd"], ax=axes[0, 0], color="skyblue")
axes[0, 0].set_title("Monthly EMI - Box Plot (Default Whisker)")
# Q-Q Plot
stats.probplot(df["monthly_emi_usd"], dist="norm", plot=axes[0, 1])
axes[0, 1].set_title("Monthly EMI - Q-Q Plot")

sn.boxplot(
    x=df["debt_to_income_ratio"], ax=axes[1, 0], whis=0.5, color="salmon"
)
axes[1, 0].set_title("Debt-to-Income - Box Plot (Whisker = 0.5)")

# Q-Q Plot
stats.probplot(df["debt_to_income_ratio"], dist="norm", plot=axes[1, 1])
axes[1, 1].set_title("Debt-to-Income - Q-Q Plot")

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import norm

mu = 0.0
sigma = 1.0
figsize,(ax1,ax2)=  plt.subplots(nrows= 2, ncols = 1, figsize= (14,10))
# fig,((ax_box, ax_hist),(ax_box2, ax_hist2)) = plt.subplots(nrows = 2, ncols =2, figsize = (14,10))
x= df["debt_to_income_ratio"].sort_values()

y = norm.pdf(x, mu,sigma)

ax1.plot(x,y, label = "normal_distribution")
sn.histplot(df['emi_to_income_ratio'], kde=True, stat = 'density', ax=ax2 ,color = 'skyblue')
ax2.set_title("distribution")
plt.title("standard normal distribution")
plt.show()
```
